## Section 2: Picking Specific Items (Indexes)

In MakeCode, if you wanted the very first item in your Array, you used the `get value at [0]` block. Python works the exact same way!

We call the position of the item its **Index**. Computers always start counting at **0**. So, the first item is at index 0, the second is at index 1, and so on.

Run the code below to equip the second item in the list.

```python
# 🟢 RUN ME: Grabbing an item by its index
weapons = ["Katana", "Nunchucks", "Bow"]

# Remember, index 1 is actually the SECOND item!
equipped = weapons[1] 
print("You have equipped the:", equipped)
```

## Section 3: Adding Randomness!

Games are boring if they are the exact same every time. In MakeCode, you probably used the purple **`pick random 0 to 10`** Math block all the time.

Python has a built-in module for this called `random`. We just have to `import` it first! We can use `random.randint()` for numbers, or `random.choice()` to let Python randomly pick an item from our list!

Run the cell below a few times to see what random loot drops.

```python
# 🟢 RUN ME: Try running this cell 3 different times!
import random

loot_table = ["Common Sword", "Rare Shield", "Legendary Armor", "Junk!"]
drop = random.choice(loot_table)

print("You defeated the monster and got:", drop)
```
