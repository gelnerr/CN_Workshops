# The Magic of Loops
In Microsoft MakeCode Arcade, if you wanted to spawn 5 enemies, you wouldn't drag out the "spawn enemy" block 5 times. Instead, you used the green **`repeat [ 5 ] times`** block and put the spawn block inside it.

In Python, we do this using a **`for`** loop and a special command called `range()`.

Run the cell below. Notice how it prints the message 3 times, but we only had to write the `print` command once!

```python
# 🟢 RUN ME: Click here and press Shift + Enter
for i in range(3):
    print("Throwing a ninja star! 🌟")
```

## Section 2: Using the 'i' in the Loop

Have you been wondering what the `i` in `for i in range(3):` means?

`i` is just a variable! Python uses it to keep track of how many times the loop has run. Think of it like a MakeCode **`index`** variable. 

By default, Python starts counting at **0**. Run the code below to see!

```python
# 🟢 RUN ME: Seeing the 'i' variable count
for i in range(5):
    print("The loop is on number:", i)
```

## Section 3: The `while` Loop

MakeCode also has a **`while [ true ] do`** block. Instead of running a specific number of times, a `while` loop keeps running *as long as a condition is true*.

It's like saying: *"While my health is greater than 0, keep fighting!"*

Run the cell below. Watch how the boss health drops until the loop stops.

```python
# 🟢 RUN ME: Defeating the Boss with a While Loop
boss_health = 3

while boss_health > 0:
    print("Boss health is", boss_health, "- Attacking!")
    boss_health = boss_health - 1  # We take away 1 health each time!

print("Boss defeated! 🎉")
```
