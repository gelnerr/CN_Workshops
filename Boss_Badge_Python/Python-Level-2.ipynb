# Making Decisions in Code
In Microsoft MakeCode Arcade, you use the green, diamond-shaped **Logic blocks** (`0 < 0` or `0 = 0`) to check if something is True or False. Then, you snap them into an `if... then` block so your game can make a decision (like *"IF player touches lava, THEN lose a life"*).

In Python, we type these out using **Conditionals**! 

First, let's look at how Python compares things. We use `==` to check if two things are exactly the same. (We use two equals signs because a single `=` is for variables!).

Run the cell below to see Python play True or False.

```python
# 🟢 RUN ME: Click here and press Shift + Enter
print("Is 10 greater than 5?", 10 > 5)
print("Does 5 equal 5?", 5 == 5)
print("Does 'Ninja' equal 'Pirate'?", "Ninja" == "Pirate")
```

## Section 2: The `if` Statement and the Mighty Colon (`:`)

Now let's use an `if` statement. In Python, `if` statements have two very important rules:
1. They must end with a colon `:`
2. The code inside the `if` block must be **indented** (pushed in with the Tab key).

This indentation is how Python knows which code belongs inside the `if` block! Run the code below to see it in action.

```python
# 🟢 RUN ME: A simple Python if statement
ninja_score = 100

if ninja_score == 100:
    print("Perfect score! You are a Python master.")
```

## Section 3: `else` and `elif` (Else If)

What if we want something different to happen when the condition is False? Just like MakeCode's `if... else` block, Python has `else`!

Python also has `elif` (which stands for "else if"). It lets us check multiple things in a row. Run the cell below and try changing the `health` variable to see what prints out.

```python
# 🟢 RUN ME: Try changing health to 50, then to 0, and run it again!
health = 100

if health == 100:
    print("You have full health!")
elif health > 0:
    print("You are hurt, find a potion!")
else:
    print("Game Over!")
```
